 Data Cleaning Summary

The original Mumbai housing dataset was inspected using pandas. Exact duplicate
rows were identified and removed, leaving 55,726 records. Column names were
standardized to lowercase with underscores. The dataset contained no missing
values after inspection. Data types, numerical variables, BHK distribution,
property types, area, price, and locality information were also examined.

The cleaned dataset was saved as `data/raw_cleaned.csv`.
